```python
def build_connection_string(db_config: dict) -> str:
    user: str = db_config["user"]
    password: str = db_config["pwd"]
    host: str = db_config["host"]
    port: int = db_config["port"]
    database: str = db_config["db"]
    return f"postgresql://{user}:{password}@{host}:{port}/{database}"


def create_afisha_engine() -> Engine:
    db_config = {
        "user": "praktikum_student",
        "pwd": "Sdf4$2;d-d30pp",
        "host": "rc1b-wcoijxj3yxfsf3fs.mdb.yandexcloud.net",
        "port": 6432,
        "db": "data-analyst-afisha",
    }

    connection_string = build_connection_string(db_config)
    engine: Engine = create_engine(connection_string, connect_args={"sslmode": "require"})
    return engine


def fetch_dataframe(query: str, engine: Engine) -> pd.DataFrame:
    return pd.read_sql_query(query, con=engine)
```

```python
sql = """
WITH set_config_precode AS (
  SELECT set_config('synchronize_seqscans', 'off', true)
)

SELECT 
    p.user_id,
    p.device_type_canonical,
    p.order_id,
    p.created_dt_msk AS order_dt,
    p.created_ts_msk AS order_ts,
    p.currency_code,
    p.revenue,
    p.tickets_count,
    p.created_dt_msk::DATE - LAG(p.created_dt_msk::DATE) OVER (
        PARTITION BY p.user_id 
        ORDER BY p.created_dt_msk
    ) AS days_since_prev,
    p.event_id,
    e.event_name_code AS event_name,
    e.event_type_main,
    p.service_name,
    c.city_name,
    r.region_name
FROM afisha.purchases AS p
JOIN afisha.events AS e
    ON p.event_id = e.event_id
JOIN afisha.city AS c
    ON e.city_id = c.city_id
JOIN afisha.regions AS r
    ON c.region_id = r.region_id
WHERE 
    p.device_type_canonical IN ('mobile', 'desktop')
    AND e.event_type_main != 'фильм'
ORDER BY p.user_id;
"""

orders_df = fetch_dataframe(sql, engine)
orders_df.head(10)
```

**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведите выручку к единой валюте — российскому рублю.

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

---


```python
url = "https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv"
rates_raw = pd.read_csv(url)

# Определяем столбцы даты и курса
date_cols = [c for c in rates_raw.columns if any(k in c.lower() for k in ["date", "dt"]) ]
rate_cols = [c for c in rates_raw.columns if c not in date_cols]

date_col = date_cols[0] if date_cols else rates_raw.columns[0]
rate_col = rate_cols[0] if rate_cols else rates_raw.columns[-1]

rates = rates_raw.copy()
rates[date_col] = pd.to_datetime(rates[date_col], errors="coerce")
rates = rates.dropna(subset=[date_col])
rates["rate_date"] = rates[date_col].dt.date

# Курс указан за 100 тенге → приводим к 1 тенге, агрегируем до 1 строки на дату
rates["rate_per_kzt"] = pd.to_numeric(rates[rate_col], errors="coerce") / 100.0
rates = rates[["rate_date", "rate_per_kzt"]].dropna()
rates = rates.groupby("rate_date", as_index=False)["rate_per_kzt"].mean()

# Соединяем по дате заказа (без размножения строк)
orders_df["order_date"] = orders_df["order_dt"].dt.date
orders_df = orders_df.merge(rates, left_on="order_date", right_on="rate_date", how="left")

# Конвертация
cur = orders_df["currency_code"].str.lower().str.strip()
orders_df["revenue_rub"] = np.where(cur == "kzt", orders_df["revenue"] * orders_df["rate_per_kzt"], orders_df["revenue"])

# Чистка служебных столбцов
orders_df = orders_df.drop(columns=[c for c in ["order_date", "rate_date", "rate_per_kzt"] if c in orders_df.columns])

# Контроль
print("Добавлен столбец revenue_rub. Пример:")
display(orders_df[["currency_code", "revenue", "revenue_rub"]].head(10))
```

Сегмент — это группа пользователей, объединённых определённым признаком, то есть объединённые принадлежностью к категории. Например, все клиенты, сделавшие первый заказ с мобильного телефона, — это сегмент.
- Ответьте на вопрос: равномерно ли распределены пользователи по сегментам или есть выраженные «точки входа» — сегменты с наибольшим числом пользователей?

```python
def segment_counts(df: pd.DataFrame, col: str) -> pd.DataFrame:
    s = df[col].fillna("NA").astype(str)
    counts = s.value_counts(dropna=False).rename("users_count")
    share = s.value_counts(dropna=False, normalize=True).rename("users_share")
    out = pd.concat([counts, share], axis=1).reset_index().rename(columns={"index": col})
    out["users_share"] = out["users_share"].round(4)
    return out.sort_values("users_count", ascending=False)
```

```python
# Попытка использовать phik; при отсутствии — запасной вариант
use_phik = False
try:
    import phik  # noqa: F401
    use_phik = True
except Exception:
    use_phik = False

if use_phik:
    corr = df.phik_matrix(interval_cols=interval_cols)
    target_col = 'total_orders'
    target_corr = corr[target_col].drop(index=target_col).sort_values(ascending=False)

    plt.figure(figsize=(8, 6))
    sns.heatmap(corr.loc[target_corr.index.tolist() + [target_col], [target_col]], annot=True, fmt='.2f', cmap='viridis')
    plt.title('phi_k с total_orders')
    plt.tight_layout()
    plt.show()

    print('Топ признаков по phi_k с total_orders:')
    print(target_corr.head(10))
else:
    # Fallback: one-hot для категорий + Спирмен с total_orders
    num_cols = [c for c in df.select_dtypes(include=['number']).columns if c != 'total_orders']
    cat_cols = [c for c in df.columns if c not in num_cols + ['total_orders']]
    df_enc = pd.concat([
        df[['total_orders'] + num_cols].copy(),
        pd.get_dummies(df[cat_cols].astype('category'), drop_first=True)
    ], axis=1)
    corr_vec = df_enc.corr(method='spearman')['total_orders'].drop('total_orders').abs().sort_values(ascending=False)

    top_feats = corr_vec.head(20)
    plt.figure(figsize=(8, 6))
    sns.barplot(x=top_feats.values, y=top_feats.index, orient='h', palette='viridis')
    plt.title('Ассоциация с total_orders (fallback)')
    plt.xlabel('|Spearman|')
    plt.tight_layout()
    plt.show()

    print('Топ признаков по ассоциации (fallback):')
    print(top_feats)
```
